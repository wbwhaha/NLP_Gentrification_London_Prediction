```python
import pandas as pd
```

```python
Tenure_2011 = pd.read_csv("../Dataset/Census 2011/Tenure 2011.csv", skiprows=8)
Occupation_2011 = pd.read_csv("../Dataset/Census 2011/Occupation 2011.csv", skiprows=9)
Ethnic_group_2011 = pd.read_csv("../Dataset/Census 2011/Ethnic group 2011.csv", skiprows=8)
Age_structure_2011 = pd.read_csv("../Dataset/Census 2011/Age structure 2011.csv", skiprows=8)
```

```python
datadict_2011 = {
    'Tenure_2011': Tenure_2011,
    'Occupation_2011': Occupation_2011,
    'Ethnic_group_2011': Ethnic_group_2011,
    'Age_structure_2011': Age_structure_2011
}
```

```python
Tenure_2021 = pd.read_csv("../Dataset/Census 2021/Tenure 2021.csv", skiprows=7)
Occupation_2021 = pd.read_csv("../Dataset/Census 2021/Occupation 2021.csv", skiprows=7)
Ethnic_group_2021 = pd.read_csv("../Dataset/Census 2021/Ethnic group 2021.csv", skiprows=7)
Age_structure_2021 = pd.read_csv("../Dataset/Census 2021/Age structure 2021.csv", skiprows=5)
```

```python
datadict_2021 = {
    'Tenure_2021': Tenure_2021,
    'Occupation_2021': Occupation_2021,
    'Ethnic_group_2021': Ethnic_group_2021,
    'Age_structure_2021': Age_structure_2021
}
```

```python
LSOA11CD = pd.read_csv("../Output/LSOA11CD.csv")
LSOA21CD = pd.read_csv("../Output/LSOA21CD.csv")
```

```python
# Loop over each dataset to perform cleaning operations
for name, df in datadict_2011.items():
    # Remove rows where the LSOA field is missing
    df.dropna(subset=['2011 super output area - lower layer'], inplace=True)

    # Split the column into LSOA code and Borough name
    df[['LSOA11CD', 'Borough']] = df['2011 super output area - lower layer'].str.split(' : ', expand=True)

    # Extract only the alphabetic part of the Borough name
    df['Borough'] = df['Borough'].str.extract(r'([A-Za-z\s]+)')

    # Filter rows to keep only those with valid LSOA codes
    df = df[df['LSOA11CD'].isin(LSOA11CD['LSOA11CD'])]

    # Drop the original raw column
    df.drop(columns=['2011 super output area - lower layer'], inplace=True)

    # Exyract the column names
    cols = df.columns.tolist()

    # Reorder: move last two columns to the front
    new_order = cols[-2:] + cols[:-2]
    df = df[new_order]

    # Update the cleaned DataFrame in the dictionary
    datadict_2011[name] = df

# Reassign the cleaned DataFrames back to their original variable names
Tenure_2011 = datadict_2011['Tenure_2011']
Occupation_2011 = datadict_2011['Occupation_2011']
Ethnic_group_2011 = datadict_2011['Ethnic_group_2011']
Age_structure_2011 = datadict_2011['Age_structure_2011']
```

```python
# Loop over each dataset to perform cleaning operations
for name, df in datadict_2021.items():
    # Remove rows where the LSOA field is missing
    df.dropna(subset=['2021 super output area - lower layer'], inplace=True)

    # Split the column into LSOA code and Borough name
    df[['LSOA21CD', 'Borough']] = df['2021 super output area - lower layer'].str.split(' : ', expand=True)

    # Extract only the alphabetic part of the Borough name
    df['Borough'] = df['Borough'].str.extract(r'([A-Za-z\s]+)')

    # Filter rows to keep only those with valid LSOA codes
    df = df[df['LSOA21CD'].isin(LSOA21CD['LSOA21CD'])]

    # Drop the original raw column
    df.drop(columns=['2021 super output area - lower layer'], inplace=True)

    # Exyract the column names
    cols = df.columns.tolist()

    # Reorder: move last two columns to the front
    new_order = cols[-2:] + cols[:-2]
    df = df[new_order]
    
    # Update the cleaned DataFrame in the dictionary
    datadict_2021[name] = df

# Reassign the cleaned DataFrames back to their original variable names
Tenure_2021 = datadict_2021['Tenure_2021']
Occupation_2021 = datadict_2021['Occupation_2021']
Ethnic_group_2021 = datadict_2021['Ethnic_group_2021']
Age_structure_2021 = datadict_2021['Age_structure_2021']
```

```python
# Define save path templates
save_path_2011 = "../Dataset/Census London 2011/{}.csv"
save_path_2021 = "../Dataset/Census London 2021/{}.csv"

# Save 2011 datasets
for name, df in datadict_2011.items():
    df.to_csv(save_path_2011.format(name), index=False)

# Save 2021 datasets
for name, df in datadict_2021.items():
    df.to_csv(save_path_2021.format(name), index=False)
```
