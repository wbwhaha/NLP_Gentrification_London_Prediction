```python
import pandas as pd
```

```python
# List of London borough names
london_boroughs = [
    "Barking and Dagenham", "Barnet", "Bexley", "Brent", "Bromley",
    "Camden", "City of London", "Croydon", "Ealing", "Enfield", "Greenwich",
    "Hackney", "Hammersmith and Fulham", "Haringey", "Harrow", "Havering",
    "Hillingdon", "Hounslow", "Islington", "Kensington and Chelsea", "Kingston upon Thames",
    "Lambeth", "Lewisham", "Merton", "Newham", "Redbridge",
    "Richmond upon Thames", "Southwark", "Sutton", "Tower Hamlets", "Waltham Forest",
    "Wandsworth", "Westminster"
    ]
```

```python
df = pd.read_csv("../Dataset/LSOA Best-Fit Lookup (2001-2011-2021)/LSOA_(2011)_to_LSOA_(2021)_to_Local_Authority_District_(2022)_Exact_Fit_Lookup_for_EW_(V3).csv")
```

```python
# Select the data for the London area 
df = df[df['LAD22CD'].str.startswith('E09')]
df.drop(columns=['LAD22NMW'], inplace=True)
```

```python
df
```

```python
counts_lsoa_2021 = df['LSOA21CD'].value_counts().reset_index()
counts_lsoa_2021.columns = ['Element', 'Count']

print(counts_lsoa_2021)
```

```python
counts_lsoa_2011 = df['LSOA11CD'].value_counts().reset_index()
counts_lsoa_2011.columns = ['Element', 'Count']

print(counts_lsoa_2011)
```

```python
counts_change_2011_2021 = df['CHGIND'].value_counts().reset_index()
counts_change_2011_2021.columns = ['Element', 'Count']

print(counts_change_2011_2021)
```

```python
df_lsoa_U_2011_2021 = df[df['CHGIND'] == 'U']
```

```python
df_lsoa_U_2011_2021
```

```python
# Select the columns needed in the analysis
selected_columns = ['LSOA11CD', 'LSOA21CD', 'LAD22CD', 'LAD22NM']
df_lsoa_U_2011_2021 = df_lsoa_U_2011_2021[selected_columns]
```

```python
df_lsoa_U_2011_2021.to_csv('../Output/lsoa_2011_to_2021_unchnaged.csv', index=False)
```

```python
df_ = pd.read_excel("../Dataset/LSOA Best-Fit Lookup (2001-2011-2021)/2001_to_2011_OA-LSOA-MSOA-WARD-BOROUGH.xls", engine='xlrd', sheet_name='LSOA')
```

```python
# The format is usually like "Camden 001A", so we extract the part before the first number
df_["borough_name"] = df_["LSOA11NM"].str.extract(r"^([A-Za-z \-&']+?)\s+\d")

# Remove any leading or trailing whitespace
df_["borough_name"] = df_["borough_name"].str.strip()

# Filter rows where the extracted borough name is in the official list
df_ = df_[df_["borough_name"].isin(london_boroughs)]

df_.drop(columns=["borough_name"], inplace=True)
```

```python
df_
```

```python
counts_2011_ = df_['LSOA11NM'].value_counts().reset_index()
counts_2011_.columns = ['Element', 'Count']

print(counts_2011_)
```

```python
counts_2001 = df_['LSOA01NM'].value_counts().reset_index()
counts_2001.columns = ['Element', 'Count']

print(counts_2001)
```

```python
counts_change_2001_2011 = df_['CHGIND'].value_counts().reset_index()
counts_change_2001_2011.columns = ['Element', 'Count']

print(counts_change_2001_2011)
```

```python
df_lsoa_U_2001_2011 = df_[df_['CHGIND'] == 'U']
```

```python
df_lsoa_U_2001_2011
```

```python
df_lsoa_U_2001_2021 = pd.merge(df_lsoa_U_2001_2011, df_lsoa_U_2011_2021, left_on='LSOA11CD', right_on='LSOA11CD', how='left')
df_lsoa_U_2001_2021.dropna(subset=['LSOA21CD'], inplace=True)
```

```python
# Select the columns needed in the analysis
selected_columns = ['LSOA01CD', 'LSOA11CD', 'LSOA21CD', 'LAD22CD', 'LAD22NM']
df_lsoa_U_2001_2021 = df_lsoa_U_2001_2021[selected_columns]
```
