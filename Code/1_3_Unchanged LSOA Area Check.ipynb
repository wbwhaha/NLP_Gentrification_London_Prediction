```python
import pandas as pd
```

```python
df = pd.read_csv("../Dataset/LSOA Best-fit lookup(2001-2011-2021)/LSOA_(2011)_to_LSOA_(2021)_to_Local_Authority_District_(2022)_Best_Fit_Lookup_for_EW_(V2).csv")
```

```python
df = df[df['LAD22CD'].str.startswith('E09')]
df.drop(columns=['LAD22NMW'], inplace=True)
```

```python
df
```

```python
# Count occurrences of each LSOA code in 2001 and 2011
count_2011 = df['LSOA11CD'].value_counts()
count_2021 = df['LSOA21CD'].value_counts()

# Filter rows where both lsoa2001cd and lsoa2011cd appear only once,
# ensuring a strict one-to-one mapping (no splits or merges)
df_lsoa_U_2011_2021 = df[(df['LSOA11CD'].map(count_2011) == 1) & (df['LSOA21CD'].map(count_2021) == 1)]

# Display the final filtered dataset
df_lsoa_U_2011_2021
```

```python
df_ = pd.read_excel("../Dataset/LSOA Best-fit lookup(2001-2011-2021)/2001_to_2011_OA-LSOA-MSOA-WARD-BOROUGH.xls", engine='xlrd', sheet_name='LSOA')
```

```python
df_ = df_.iloc[:4853]
```

```python
df_lsoa_U_2001_2011 = df_[df_['CHGIND'] == 'U']
```

```python
df_lsoa_U_2001_2021 = pd.merge(df_lsoa_U_2001_2011, df_lsoa_U_2011_2021, left_on='LSOA11CD', right_on='LSOA11CD', how='left')
```
