```python
# Directory containing CSV files
data_dir_2021 = "../Dataset/Census London 2021/"

merged_df_2021 = None

# Loop over all files in the directory
for filename in os.listdir(data_dir_2021):
    if filename.endswith("2021.csv") and not filename.startswith(datasets_used_in_classification):

        filepath = os.path.join(data_dir_2021, filename)
        
        df_2021 = pd.read_csv(filepath)

        # If this is the first file, initialize merged_df
        if merged_df_2021 is None:
            merged_df_2021 = df_2021

        else:
            # Drop duplicate columns before merging
            df_2021 = df_2021[[col for col in df_2021.columns if col == 'LSOA21CD' or col not in merged_df_2021.columns]]
            # Merge on 'LSOACD21'
            merged_df_2021 = pd.merge(merged_df_2021, df_2021, on='LSOA21CD', how='outer')
```

```python
merged_df_2021['Economically active: Employee: Part-time'] = (
    merged_df_2021['Economically active (excluding full-time students): In employment: Employee: Part-time'] +
    merged_df_2021['Economically active and a full-time student: In employment: Employee: Part-time']
)

merged_df_2021['Economically active: Employee: Full-time'] = (
    merged_df_2021['Economically active (excluding full-time students): In employment: Employee: Full-time'] +
    merged_df_2021['Economically active and a full-time student: In employment: Employee: Full-time']
)

merged_df_2021['Economically active: Self-employed with employees: Part-time'] = (
    merged_df_2021['Economically active (excluding full-time students): In employment: Self-employed with employees: Part-time'] +
    merged_df_2021['Economically active and a full-time student: In employment: Self-employed with employees: Part-time']
)

merged_df_2021['Economically active: Self-employed with employees: Full-time'] = (
    merged_df_2021['Economically active (excluding full-time students): In employment: Self-employed with employees: Full-time'] +
    merged_df_2021['Economically active and a full-time student: In employment: Self-employed with employees: Full-time']
)

merged_df_2021['Economically active: Self-employed without employees: Part-time'] = (
    merged_df_2021['Economically active (excluding full-time students): In employment: Self-employed without employees: Part-time'] +
    merged_df_2021['Economically active and a full-time student: In employment: Self-employed without employees: Part-time']
)

merged_df_2021['Economically active: Self-employed without employees: Full-time'] = (
    merged_df_2021['Economically active (excluding full-time students): In employment: Self-employed without employees: Full-time'] +
    merged_df_2021['Economically active and a full-time student: In employment: Self-employed without employees: Full-time']
)

merged_df_2021['Economically active: Unemployed'] = (
    merged_df_2021['Economically active (excluding full-time students): Unemployed'] +
    merged_df_2021['Economically active and a full-time student: Unemployed']
)

student_cols = [col for col in merged_df_2021.columns if col.startswith('Economically active and a full-time student')]
merged_df_2021['Economically active: Full-time student'] = merged_df_2021[student_cols].sum(axis=1)

# Economically active: Total
active_cols = [col for col in merged_df_2021.columns if col.startswith('Economically active') and 'inactive' not in col.lower()]
merged_df_2021['Economically active: Total'] = merged_df_2021[active_cols].sum(axis=1)

# Economically inactive: Total
inactive_cols = [col for col in merged_df_2021.columns if col.startswith('Economically inactive')]
merged_df_2021['Economically inactive: Total'] = merged_df_2021[inactive_cols].sum(axis=1)

eco_act_drop = [
    'Economically active (excluding full-time students)',
    'Economically active (excluding full-time students):In employment',
    'Economically active (excluding full-time students):In employment:Employee',
    'Economically active (excluding full-time students): In employment: Employee: Part-time',
    'Economically active (excluding full-time students): In employment: Employee: Full-time',
    'Economically active (excluding full-time students):In employment:Self-employed with employees',
    'Economically active (excluding full-time students): In employment: Self-employed with employees: Part-time',
    'Economically active (excluding full-time students): In employment: Self-employed with employees: Full-time',
    'Economically active (excluding full-time students):In employment:Self-employed without employees',
    'Economically active (excluding full-time students): In employment: Self-employed without employees: Part-time',
    'Economically active (excluding full-time students): In employment: Self-employed without employees: Full-time',
    'Economically active (excluding full-time students): Unemployed',
    'Economically active and a full-time student',
    'Economically active and a full-time student:In employment',
    'Economically active and a full-time student:In employment:Employee',
    'Economically active and a full-time student: In employment: Employee: Part-time',
    'Economically active and a full-time student: In employment: Employee: Full-time',
    'Economically active and a full-time student:In employment:Self-employed with employees',
    'Economically active and a full-time student: In employment: Self-employed with employees: Part-time',
    'Economically active and a full-time student: In employment: Self-employed with employees: Full-time',
    'Economically active and a full-time student:In employment:Self-employed without employees',
    'Economically active and a full-time student: In employment: Self-employed without employees: Part-time',
    'Economically active and a full-time student: In employment: Self-employed without employees: Full-time',
    'Economically active and a full-time student: Unemployed',
]

merged_df_2021.drop(columns=eco_act_drop, inplace=True)
```
