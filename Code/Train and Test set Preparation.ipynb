```python
import os
import pandas as pd
```

```python
# Exclude the datasets used in the gentrification classification
datasets_used_in_classification = ("Education", "Ethnic", "NS-SeC", "Occupation", "Tenure")
```

```python
# Directory containing CSV files
data_dir_2011 = "../Dataset/Census London 2011/"

merged_df_2011 = None

# Loop over all files in the directory
for filename in os.listdir(data_dir_2011):
    if filename.endswith("mapped.csv") and not filename.startswith(datasets_used_in_classification):

        filepath = os.path.join(data_dir_2011, filename)
        
        df = pd.read_csv(filepath)

        # If this is the first file, initialize merged_df
        if merged_df_2011 is None:
            merged_df_2011 = df

        else:
            # Drop duplicate columns before merging
            df = df[[col for col in df.columns if col == 'LSOA21CD' or col not in merged_df_2011.columns]]
            # Merge on 'LSOACD21'
            merged_df_2011 = pd.merge(merged_df_2011, df, on='LSOA21CD', how='outer')

merged_df_2011.columns = [merged_df_2011.columns[0]] + [col.replace('_weighted', '') for col in merged_df_2011.columns[1:]]
```
