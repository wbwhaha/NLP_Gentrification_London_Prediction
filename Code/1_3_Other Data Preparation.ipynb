```python
import pandas as pd
```

```python
df_num = pd.read_excel("../Dataset/Other Datasets/HPSSA Dataset 41 - Number of residential property sales by LSOA.xls", engine='xlrd', sheet_name='Data', skiprows=5)
```

```python
# Select the data of year 2011 and 2021
df_num = df_num.iloc[:, :4].join(df_num.filter(regex="(2011|2021)$"))
```

```python
# Select the data in London area
df_num = df_num[df_num['Local authority code'].str.startswith('E09')]
```

```python
df_num
```

```python
df_num.columns
```

```python
columns_to_check = ['Year ending Mar 2011', 'Year ending Jun 2011',
       'Year ending Sep 2011', 'Year ending Dec 2011', 'Year ending Mar 2021',
       'Year ending Jun 2021', 'Year ending Sep 2021', 'Year ending Dec 2021']

counts = df_num[columns_to_check].apply(lambda col: (col == 0).sum())

print(counts)
```

```python
df_num['Number of sales 2011'] = df_num[['Year ending Mar 2011', 'Year ending Jun 2011', 'Year ending Sep 2011', 'Year ending Dec 2011']].sum(axis=1)
df_num.drop(columns=['Year ending Mar 2011', 'Year ending Jun 2011', 'Year ending Sep 2011', 'Year ending Dec 2011'], inplace=True)

df_num['Number of sales 2021'] = df_num[['Year ending Mar 2021', 'Year ending Jun 2021', 'Year ending Sep 2021', 'Year ending Dec 2021']].sum(axis=1)
df_num.drop(columns=['Year ending Mar 2021', 'Year ending Jun 2021', 'Year ending Sep 2021', 'Year ending Dec 2021'], inplace=True)
```

```python
df_num.rename(columns={'LSOA code': 'LSOA11CD'}, inplace=True)
```

```python
# Save file
df_num.to_csv("../Dataset/Other Datasets/Number of sales.csv", index=False)
```

```python
# List of London borough names
london_boroughs = [
    "Barking and Dagenham", "Barnet", "Bexley", "Brent", "Bromley",
    "Camden", "City of London", "Croydon", "Ealing", "Enfield", "Greenwich",
    "Hackney", "Hammersmith and Fulham", "Haringey", "Harrow", "Havering",
    "Hillingdon", "Hounslow", "Islington", "Kensington and Chelsea", "Kingston upon Thames",
    "Lambeth", "Lewisham", "Merton", "Newham", "Redbridge",
    "Richmond upon Thames", "Southwark", "Sutton", "Tower Hamlets", "Waltham Forest",
    "Wandsworth", "Westminster"
    ]
```

```python
df_pop_2011 = pd.read_csv("../Dataset/Other Datasets/census_population_2011.csv", skiprows=8)
```

```python
import re

# Create a regular expression by joining all London boroughs, which will be used for matching
regex = r'\b(?:' + '|'.join(re.escape(b) for b in london_boroughs) + r')\b'

# Filter rows where the 'area_info' column contains any of the London boroughs
df_pop_london_2011 = df_pop_2011[df_pop_2011['2011 super output area - lower layer'].str.contains(regex, case=False, na=False, regex=True)]

# Extract LSOA code, assuming it's the part after the colon ':'
df_pop_london_2011['2011 super output area - lower layer'] = df_pop_london_2011['2011 super output area - lower layer'].str.split(':').str[0].str.strip()

df_pop_london_2011.rename(columns={'2011 super output area - lower layer': 'LSOA11CD'}, inplace=True)

# Display the resulting dataframe
df_pop_london_2011 = df_pop_london_2011.iloc[:,:2]

df_pop_london_2011.rename(columns={'All usual residents':'All residents'}, inplace=True)
```

```python
df_pop_london_2011
```

```python
# Save file
df_pop_london_2011.to_csv("../Output/Population_2011.csv", index=False)
```
